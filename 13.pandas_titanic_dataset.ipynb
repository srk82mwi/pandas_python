**Series - 1D array**

```python
import pandas as pd

marks = pd.Series([10,45,34,62,29,43.5,"pakistan"])
print(marks)
print(marks[6])
```

```python
print(marks.index)
```

```python
result = pd.Series([10,45,34,62,29,43], index=["Ali","Sara","Farooq","Ahmad","Javed","Imran"])
print(result)
print(result['Farooq'])
```

```python
print(result.index)
print(result.values)
```

**Series using Dictionaries**

```python
st_marks = pd.Series({"Ali":10,"Sara":45,"Farooq":34,"Ahmad":62,"Javed":29,"Imran":43})
print(st_marks)
print(st_marks['Ahmad'])
print(st_marks.iloc[3])
```

```python
print(st_marks.iloc[1:3])
```

```python
print(st_marks[['Farooq','Javed']]+5)
```

**DataFrame  - Rows and Columns** <br>
- A **DataFrame** is a full table


```python
students = pd.DataFrame({
            "name":["Ali","Ahmad","Usman","Javed","Farooq"],
            "age":[17,19,24,21,18],
            "city":["karachi","lahore","multan","islamabad","quetta"],
            "marks":[45,54,56,43,67]
    })

print(students)

#to retrieve a column
print(students['city'])

#to retrieve a row
print(students.iloc[0])
```

```python
print(f"Dataframe has {students.shape[0]} rows and {students.shape[1]} columns")
print(students.ndim)
print(students.size)
print(students.dtypes)
print(students.columns)
print(students.index)
```

```python
r,c = students.shape
print(f"{r} rows {c} columns")
print(students.info())
```

**Selecting Columns** <br>
- df['age'] - one set of square brackets - returns a series
- df[['age','fare']] - two sets - returns a dataframe

```python
ages = ds_titanic['age']
print(ages.head())
print(f"Ages Mean: {round(ages.mean(),1)}")
print(f"Ages Median: {round(ages.median(),1)}")
print(f"Ages Max: {round(ages.max(),1)}")
print(f"Ages Min: {round(ages.min(),1)}")
```

```python
over_30 = ages>30
males = ds_titanic['sex']=='male'
print(f"Males: {males.sum()}")
print(f"Over 30: {over_30.sum()}")
print(ds_titanic['sex'])

#find over 30 males
over_30_males = males & over_30
print(f"Over 30 Males: {over_30_males.sum()}")
print(f"Over 30 Males: {ds_titanic['sex'][over_30_males]}")
```

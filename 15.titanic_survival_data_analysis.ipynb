```python
import numpy as np
import pandas as pd
import seaborn as sb

titanic = sb.load_dataset('titanic')
titanic
```

```python
#How many people survived overall?

titanic['survived'].sum()
sc=titanic.survived.sum()
print(f"{sc}")
s = titanic['survived'].mean()
print(f"{s:.1%}")
```

```python
titanic.groupby('sex')['survived'].mean()
```

```python
#fare bands, using filtering
cheap = titanic[titanic['fare']<10]
print(cheap['fare'].shape[0])

mid = titanic[titanic['fare'].between(10,50)]
print(mid.shape[0])

expensive = titanic[titanic['fare']>50]
expensive.shape[0]
```

```python
subject_cols = [
    "18th_century_english_novel",
    "islamic_studies",
    "understanding_holy_quran_1",
    "history_of_english_literature",
    "intro_to_literary_devices_and_genres",
    "functional_english",
    "applications_of_ict",
]

# 1. Reshape dataframe from wide to long format
df_long = bs_result.melt(
    id_vars=["student_name"],
    value_vars=subject_cols,
    var_name="subject",
    value_name="marks",
)
print(df_long)
# 2. Filter subjects where marks are less than 50 (Fail)
failed_df = df_long[df_long["marks"] < 50]

# 3. Group by student name and join subject names
result = (
    failed_df.groupby("student_name")["subject"]
    .apply(lambda x: ", ".join(x))
    .reset_index()
)

# Rename columns for clarity
result.columns = ["Student Name", "Failed Subject(s)"]

print(result.to_string(index=False))
```

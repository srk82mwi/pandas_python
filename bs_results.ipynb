```python
# print(len(bs_result.columns))
# print(len(bs_result.columns))
data =[]
d_index=[]
col = bs_result.columns
print("Subjects|                          |Average Marks|")
for m in range(2,9):
  # print(f"{col[m]} {round(bs_result[col[m]].mean(),2)}")
  data.append(round(bs_result[col[m]].mean(),2))
  d_index.append(col[m])

sub = pd.Series(data,index=d_index)
print(sub.sort_values(ascending=False))
```

```python
data =[]
d_index=[]
col = bs_result.columns
print("__________|Subjects|_________________| Failed Students|_____")
for m in range(2,9):
  # print(f"{col[m]} {round(bs_result[col[m]].mean(),2)}")
  fa = bs_result[bs_result[col[m]]<50]
  data.append(fa.shape[0])
  d_index.append(col[m])

sub = pd.Series(data,index=d_index)
print(sub.sort_values(ascending=False))
```

0h:41m <br>
`.isin()` - a shortcut for many OR's <br>
Writing (x=='a') | (x=='b') | (x=='c') gets ugly fast <br>
isin() does it cleanly.


```python
# the long way
long_way = titanic[(titanic['pclass'] == 1) | (titanic['pclass'] ==2)]

# the clean way
clean_way = titanic[titanic['pclass'].isin([1,2])]
clean_way.head(15)
print(long_way.shape[0])
print(clean_way.shape[0])
```

**.between()** - a shortcut for ranges


```python
# the long way
# passenger between 13 and 19 years of age

long_way = titanic[(titanic['age']>=13) & (titanic['age']<=19)]
print(long_way.shape[0])

# clean way
clean_way = titanic[titanic['age'].between(13,19)]
print(clean_way.shape[0])
```

```python
women = titanic[titanic['sex']=='female']
men = titanic[titanic['sex']=='male']
print(round(women['survived'].mean(),2))
print(round(men['survived'].mean(),2))
```

```python
# find the average fare of each class
for c in [1,2,3]:
    cl = titanic['pclass']==c
    cl_fare = titanic[cl]['fare'].mean()
    print(f"Class {c} avg. fare: {round(cl_fare,2)}")

for c in [1,2,3]:
    cl= titanic[titanic['pclass']==c]
    cl_fare = cl['fare'].mean()
    print(f"Class {c} avg. fare: {round(cl_fare,2)}")
```

```python
# find the average age of passengers in each class

for c in [1,2,3]:
    cl = titanic[titanic['pclass']==c]
    print(f"Avg. age of class {c}: {round(cl['age'].mean(),2)}")
```

`.nlargest()` and `.nsmallest()` -  a shortcut <br>
when you only want the top N, this is shorter and faster than sorting the whole table.

```python
# 5 most expensive tickets
titanic.nlargest(5, "fare")
```

```python
titanic.sort_values(by="fare", ascending=False).head()
```

```python
#5 youngest passengers
titanic.nsmallest(5,"age")
```

**Adding, Renaming and Dropping columns** <br>
Adding a column <br>
just assign a new name. The calculation happens for every row at once


```python
# work on a copy so we don't damage the original.
titanic_copy = titanic.copy()
titanic_copy
titanic_copy.columns
```

```python
titanic_copy['family_size'] = titanic_copy['sibsp'] + titanic_copy['parch'] + 1
titanic_copy.columns
titanic_copy.head(10)
titanic_copy.nlargest(10,"family_size")[['survived','pclass','sex','age','sibsp','parch','family_size']]
```

```python
print(titanic_copy.columns)
print(titanic.columns)
```

why `.copy()`? <br>
if you write df = titanic you do not get a new table. you get a second name pointing at the same table. Chnage one and the other changes too.
<br>
.copy() makes a real, seperate copy. use it whenever you plan to modify a DataFrame.

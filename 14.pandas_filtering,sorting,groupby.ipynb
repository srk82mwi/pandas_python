**Boolean Filtering**

```python
bool_over_30 = titanic['age']>30
print(bool_over_30)
print(bool_over_30.sum())
```

```python
titanic[bool_over_30]
```

```python
#return males belonging to class 1
class1 = titanic['pclass'] == 1
males = titanic['sex']=='male'

class1_males = class1 & males
# titanic[class1][males]
# print(titanic[class1_males])

# filter class1, males, age<20
age_below_20 = titanic['age']<20

class1_male_age_below_20 = class1_males & age_below_20
print(titanic[class1_male_age_below_20])
print(class1_male_age_below_20.sum())
```

```python
mask = titanic['age']>30
print(titanic[mask].shape[0])
print(mask.sum())

male = titanic['sex']=='male'
print(male.sum())
print(titanic[male].shape[0])
print(titanic[male])
```

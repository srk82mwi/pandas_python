```python
import numpy as np
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset('titanic')
titanic
```

```python
titanic.info()
```

```python
isn_mean = round(titanic.isnull().mean()*100,1)
isn = titanic.isnull().sum()
repor = pd.DataFrame({
   "missing_count": isn[isn>0],
   "missing_percentage": isn_mean[isn_mean>0]
})
repor
```

```python
missing_sum = titanic.isnull().sum()

report = pd.DataFrame({
    "missing_count": titanic.isnull().sum(),
    "missing (%)":round(titanic.isnull().mean()*100,1)
})

report[report['missing_count']>0]
```

```python
clean_titanic = titanic.copy()
clean_titanic.info()
print(clean_titanic.shape)
print(len(clean_titanic))
```

```python
print(f"Before: {len(clean_titanic)}")
post_dropna = clean_titanic.dropna()
print(f"After: {len(post_dropna)}")
print(f"Lost: {len(clean_titanic) - len(post_dropna)}")
print(len(post_dropna)/len(clean_titanic)*100)
post_dropna.sample(15)
```

**`subset=`  only care about specific columns**

```python
post_dropna = clean_titanic.dropna(subset=['embarked','embark_town'])
len(post_dropna)
```

**`how='any'` vs `how='all'`**

```python
drop_any = clean_titanic.dropna(how='any')
drop_any
drop_how_all = clean_titanic.dropna(how='all')
drop_how_all
```

**`axis=1`  -   drop columns instead of rows**

```python
drop_col = clean_titanic.dropna(axis=1)
print(len(clean_titanic.columns))
len(drop_col.columns)
```

5. **List comprehension**

```python
[c for c in clean_titanic.columns if c not in drop_col]
```

5.1 **Regular way / Alternate method for List comprehension**

```python
dc = []
for c in clean_titanic.columns:
    if c not in drop_col.columns:
        dc.append(c)

print(f"Columns removed: {dc}")
```
